```python
#2 sum value
# in a given array find the pair of values which addition should match target
#for example 4+2 = 6 , 6 is target
# consider only 1 pair is there in the list which addition provide target
l = [1,2,3,5]
target = 6
```

```python
hash = {}
for i in l:
  to_check = target - i
  if to_check in hash:
    print( [to_check, i])
  else:
    hash[i] = 1

```
